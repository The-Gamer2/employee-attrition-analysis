## 2. Initial Data Exploration


```python
# Visualization: Attrition by Department
plt.figure(figsize=(8,5))
sns.countplot(data=df, x="Department", hue="Attrition")
plt.title("Attrition by Department")
plt.xlabel("Department")
plt.ylabel("Count")
plt.xticks(rotation=0)
plt.legend(title="Attrition")
plt.tight_layout()
plt.show()
```

 **Interpretation: Attrition by Department**

This chart shows the absolute number of employees who have attrited 'Yes' and not attrited 'No' within each department. We can observe that the Research & Development department has the highest number of employees overall, and consequently, the highest number of both attrited and non-attrited employees. The Sales department also shows a significant number of attritions. The Human Resources department has the fewest employees and the lowest absolute number of attritions.

```python
# Visualization: Attrition by Job Role
plt.figure(figsize=(12,6))
sns.countplot(data=df, y="JobRole", hue="Attrition", order=df["JobRole"].value_counts().index)
plt.title("Attrition by Job Role")
plt.xlabel("Count")
plt.ylabel("Job Role")
plt.legend(title="Attrition")
plt.tight_layout()
plt.show()
```

**Interpretation: Attrition by Job Role**

This visualization breaks down attrition by specific job roles. 'Sales Executive' and 'Laboratory Technician' roles show a considerable number of employees who have attrited. This indicates that these roles might have higher turnover rates or specific challenges that lead employees to leave the company. Roles like 'Manager' and 'Research Director' appear to have relatively lower attrition numbers.

```python
# Visualization: Attrition by Overtime
plt.figure(figsize=(7,5))
sns.countplot(data=df, x="OverTime", hue="Attrition")
plt.title("Attrition by Overtime")
plt.xlabel("OverTime")
plt.ylabel("Count")
plt.tight_layout()
plt.show()
```

**Interpretation: Attrition by Overtime**

This chart clearly illustrates the relationship between working overtime and attrition. Employees who work overtime 'Yes' show a significantly higher number of attritions compared to those who do not 'No'. This suggests that working overtime could be a major factor contributing to employees' decision to leave the company.

```python
# Visualization: Age Distribution of Employees
plt.figure(figsize=(8,5))
sns.histplot(data=df, x="Age", bins=20, kde=True)
plt.title("Age Distribution of Employees")
plt.xlabel("Age")
plt.ylabel("Count")
plt.tight_layout()
plt.show()
```

**Interpretation: Age Distribution of Employees**

This histogram displays the distribution of employee ages. The distribution appears to be skewed towards younger age groups, with a peak around the late 20s to early 30s. There are fewer employees in the older age brackets, indicating a relatively young workforce within the company.

```python
# Visualization: Monthly Income Distribution
plt.figure(figsize=(8,5))
sns.histplot(data=df, x="MonthlyIncome", bins=30, kde=True)
plt.title("Monthly Income Distribution")
plt.xlabel("Monthly Income")
plt.ylabel("Count")
plt.tight_layout()
plt.show()
```

**Interpretation: Monthly Income Distribution**

This histogram shows the distribution of monthly incomes among employees. The distribution is right-skewed, indicating that a majority of employees have lower to mid-range monthly incomes, with a long tail extending to much higher incomes. This suggests a varied compensation structure across different job levels and roles.

```python
# Visualization: Monthly Income by Attrition
plt.figure(figsize=(8,5))
sns.boxplot(data=df, x="Attrition", y="MonthlyIncome")
plt.title("Monthly Income by Attrition")
plt.xlabel("Attrition")
plt.ylabel("Monthly Income")
plt.tight_layout()
plt.show()
```

**Interpretation: Monthly Income by Attrition**

This box plot compares the monthly income of employees who attrite 'Yes' versus those who do not 'No'. It reveals that employees who leave the company generally have lower monthly incomes. The median monthly income for attrited employees is noticeably lower, and their income distribution tends to be concentrated at the lower end compared to non-attrited employees.

```python
# Visualization: Age by Attrition
plt.figure(figsize=(8,5))
sns.boxplot(data=df, x="Attrition", y="Age")
plt.title("Age by Attrition")
plt.xlabel("Attrition")
plt.ylabel("Age")
plt.tight_layout()
plt.show()
```

**Interpretation: Age by Attrition**

This box plot illustrates the age distribution for employees who attrite 'Yes' and those who do not 'No'. The plot indicates that younger employees, particularly those in their late 20s and early 30s, are more prone to attrition. The median age for attrited employees is lower than for non-attrited employees, suggesting that age plays a role in an employee's decision to leave.

```python
# Visualization: Overall Attrition Distribution
attrition_counts = df["Attrition"].value_counts()

plt.figure(figsize=(6,6))
plt.pie(attrition_counts, labels=attrition_counts.index, autopct="%1.1f%%", startangle=90, colors=["#66b3ff", "#ff9999"])
plt.title("Overall Attrition Distribution")
plt.tight_layout()
plt.show()
```

 **Interpretation: Overall Attrition Distribution**

This pie chart provides a clear overview of the overall attrition rate in the company. It shows that approximately 16.1% of employees have attrited 'Yes', while the vast majority, 83.9%, have not 'No'. This highlights the class imbalance in the target variable, which is an important consideration for any predictive modeling tasks.
